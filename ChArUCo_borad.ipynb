```python

import cv2

# 1. Select the ArUco dictionary
dictionary = cv2.aruco.getPredefinedDictionary(
    cv2.aruco.DICT_6X6_250
)

# 2. Create the ChArUco board
board = cv2.aruco.CharucoBoard(
    (5, 7),       # 5 columns × 7 rows of chessboard squares
    0.3,         # Chessboard square side length: 4 cm
    0.2,         # ArUco marker side length: 2 cm
    dictionary
)

# 3. Generate the board image
image = board.generateImage(
    (1200, 1600),
    marginSize=40
)

# 4. Save the image
cv2.imwrite("/home/nuo/图片/charuco_board.png", image)
```
